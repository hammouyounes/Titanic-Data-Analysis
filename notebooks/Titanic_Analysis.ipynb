Phase 3 - Feature Engineering

1 - Créer la colonne FamilySize

2. Comprendre la formule

FamilySize = SibSp + Parch + 1

Pourquoi ajouter 1 ?

Parce que le passager lui-même doit être comptabilisé.

```python
df_fe = df_median.copy()
```

```python
df_fe["FamilySize"] = df_fe["SibSp"] + df_fe["Parch"] + 1
df_fe[["SibSp", "Parch", "FamilySize"]].head(10)
```

2 - Créer la colonne IsAlone

```python
df_fe["IsAlone"] = (df_fe["FamilySize"] == 1).astype(int)
df_fe[["FamilySize", "IsAlone"]].head(10)
```

3 - Créer la colonne AgeGroup

```python
df_fe["AgeGroup"] = pd.cut(
    df_fe["Age"],
    bins=[0, 12, 19, 59, float("inf")],
    labels=["Child", "Teenager", "Adult", "Senior"],
    include_lowest=True
)
```

```python
df_fe[["Age", "AgeGroup"]].head(15)
```

Phase 4 - Statistical Analysis avec NumPy

```python
import numpy as np
```

2. Convertir la colonne Age en tableau NumPy

```python
age_array = df["Age"].to_numpy()

print(age_array)
```

3. Calculer l'âge moyen

```python
age_mean = np.nanmean(age_array)

print("Average age:", age_mean)
```

4. Calculer l'âge médian

```python
age_median = np.nanmedian(age_array)

print("Median age:", age_median)
```

5. Calculer l'écart-type

```python
age_std = np.nanstd(age_array)

print("L'écart-type : ", age_std)
```

6. Trouver l'âge minimum et maximum

```python
age_min = np.nanmin(age_array)
age_max = np.nanmax(age_array)

print("Minimum age : ",age_min)
print("Maximum age : ",age_max)
```

7. Identifier les passagers âgés de plus de 60 ans

```python
old_passengers = df[df["Age"] > 60]
old_passengers[["Name", "Age", "Sex", "Survived"]]
```

8. Remplacer les valeurs manquantes avec NumPy

```python
age_median = np.nanmedian(age_array)

age_filled = np.where(
    np.isnan(age_array),
    age_median,
    age_array
)
```

```python
df_numpy = df.copy()

df_numpy["Age"] = age_filled

df_numpy["Age"].isnull().sum()
```

9. Créer une catégorie selon l'âge

```python
df_numpy["AgeGroup"] = pd.cut(
    df_numpy["Age"],
    bins=[0, 12, 19, 59, float("inf")],
    labels=["Child", "Teenager", "Adult", "Senior"],
    include_lowest=True
)
```

```python
df_numpy[["Name", "Age", "AgeGroup"]].head(10)
```

10. Calculer le prix moyen des billets

```python
fare_array = df["Fare"].to_numpy()

fare_mean = np.mean(fare_array)

print("Average ticket fare", fare_mean)
```

11. Normaliser la colonne Fare

```python
fare_min = np.min(fare_array)
fare_max = np.max(fare_array)

fare_normalized = (
    (fare_array - fare_min) /
    (fare_max - fare_min)
)

df_numpy["Fare_normalized"] = fare_normalized
```

```python
df_numpy[["Fare", "Fare_normalized"]].head(10)
```

Phase 4 - Exploratory Data Analysis

1 : Les femmes avaient-elles un taux de survie supérieur aux hommes ?

```python
(df.groupby("Sex")["Survived"].mean() * 100).round(2)
```

Les femmes avaient un taux de survie nettement supérieur à celui des hommes. Environ 74,2 % des femmes ont survécu, contre seulement 18,89 % des hommes.
Cette observation suggère que le sexe était fortement associé aux chances de survie lors du naufrage du Titanic.

2 : Le taux de survie était-il plus élevé en première classe ?

```python
(df.groupby("Pclass")["Survived"].mean() * 100).round(2)
```

Le taux de survie diminue selon la classe du passager. Les passagers de première classe avaient davantage de chances de survivre que ceux de deuxième et troisième classe.

3 : Quelle est la tranche d'âge la plus représentée ?

```python
df_numpy["AgeGroup"].value_counts()
```

La tranche d'âge la plus représentée est celle des adultes (**Adult**, de 20 à 59 ans) avec 701 passagers (environ 78,7 % de l'effectif total). Viennent ensuite les adolescents (**Teenager**), les enfants (**Child**) et enfin les seniors (**Senior**).

4 : Quelle est la distribution des prix des billets ?

5 : Les personnes voyageant seules avaient-elles moins de chances de survivre ?

```python
(df_fe.groupby("IsAlone")["Survived"].mean() * 100).round(2)
```

Oui, les personnes voyageant seules avaient nettement moins de chances de survivre. Le taux de survie des passagers seuls (**IsAlone = 1**) était d'environ **30,35 %**, contre **50,56 %** pour les personnes accompagnées (**IsAlone = 0**).

6 : Taux de survie selon le port d'embarquement

```python
(df.groupby("Embarked")["Survived"].mean() * 100).round(2)
```

Les passagers ayant embarqué à **Cherbourg (C)** affichent le taux de survie le plus élevé (**55,36 %**), contre **38,96 %** pour **Queenstown (Q)** et **33,70 %** pour **Southampton (S)**.
Cet écart s'explique notamment par une proportion plus importante de passagers voyageant en première classe au départ de Cherbourg.

7 : Survie selon les tranches d'âge

```python
(df_numpy.groupby("AgeGroup")["Survived"].mean() * 100).round(2)
```

Les enfants (**Child**, <= 12 ans) ont bénéficié du taux de survie le plus important (**57,97 %**), ce qui illustre l'application de la règle « les femmes et les enfants d'abord ».
Le taux de survie diminue ensuite avec l'âge : **41,05 %** pour les adolescents (**Teenager**), **36,52 %** pour les adultes (**Adult**) et seulement **26,92 %** pour les seniors (**Senior**).

8 : Relation entre le prix du billet et la survie

```python
df.groupby("Survived")["Fare"].mean().round(2)
```

Le prix moyen du billet payé par les survivants (**48,40 $**) était plus de deux fois supérieur à celui des non-survivants (**22,12 $**).
Il existe une relation positive nette entre le tarif du billet et les chances de survie, ce qui confirme l'avantage procuré par les classes supérieures qui ont été évacuées en priorité.

9 : Survie selon le sexe et la classe simultanément

```python
(df.groupby(["Sex", "Pclass"])["Survived"].mean() * 100).round(2)
```

L'analyse simultanée du sexe et de la classe révèle des disparités saisissantes :
- **Femmes** : la survie était quasi-totale en 1ère classe (**96,81 %**) et 2ème classe (**92,11 %**), mais chutait à **50,00 %** en 3ème classe.
- **Hommes** : seuls les hommes de 1ère classe ont atteint un taux notable (**36,89 %**), contre seulement **15,74 %** en 2ème classe et **13,54 %** en 3ème classe.
Le sexe a constitué le premier facteur discriminant, amplifié très nettement par la classe sociale.

10 : Comparaison des distributions des âges entre survivants et non-survivants

```python
df.groupby("Survived")["Age"].describe().round(2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df[df["Survived"] == 1]["Age"].dropna(), bins=20, alpha=0.5, label="Survivants (1)", color="green")
plt.hist(df[df["Survived"] == 0]["Age"].dropna(), bins=20, alpha=0.5, label="Non-survivants (0)", color="red")
plt.xlabel("Âge")
plt.ylabel("Nombre de passagers")
plt.title("Distribution des âges : Survivants vs Non-survivants")
plt.legend()
plt.show()
```

L'âge médian est identique (**28 ans**) dans les deux groupes, mais l'âge moyen des survivants (**28,34 ans**) est légèrement plus bas que celui des non-survivants (**30,63 ans**).
L'examen de la distribution montre surtout un taux de survie particulièrement élevé chez les enfants de moins de 5 ans, alors que les jeunes adultes (20 à 35 ans) constituent la majorité absolue des victimes.
